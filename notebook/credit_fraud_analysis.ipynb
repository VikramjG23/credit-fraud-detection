# Credit Card Fraud Detection — Risk Analytics Project

**Objective:** Analyze transaction data to identify patterns associated with fraud, 
and build classification models to detect fraudulent transactions.

**Dataset:** Kaggle Credit Card Fraud Detection dataset — 284,807 transactions, 
492 labeled as fraud (0.17%). Features V1-V28 are PCA-transformed for privacy; 
Time and Amount are the only original, interpretable features.

**Approach:** Exploratory analysis (SQL/Excel) → Logistic Regression baseline → 
Random Forest → precision/recall tradeoff evaluation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import recall_score,f1_score,precision_score,classification_report
```

```python
creadit_df = pd.read_csv("creditcard.csv")
```

```python
creadit_df.head()
creadit_df["Class"].value_counts()
creadit_df.isnull().sum()
```

## Train/Test Split

Using `stratify=Y` to preserve the 0.17% fraud ratio in both train and test sets — 
without this, random splitting could unevenly distribute the rare fraud class.

```python
X = creadit_df.drop("Class",axis=1)
Y = creadit_df["Class"]

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

## Feature Scaling

V1-V28 are already PCA-transformed and on a similar numeric scale. Time and Amount 
are on very different scales (Amount up to ~25,000, Time in raw seconds), which 
would distort Logistic Regression's weighting. Standardizing both fixes this.

Scaler is fit only on training data and applied to test data with `.transform()` 
(not `.fit_transform()`) to avoid data leakage.

```python
scaler = StandardScaler()

X_train[['Time', 'Amount']] = scaler.fit_transform(X_train[['Time', 'Amount']])
X_test[['Time', 'Amount']] = scaler.transform(X_test[['Time', 'Amount']])
```

## Model 1: Logistic Regression (Baseline)

Using `class_weight='balanced'` to address the severe class imbalance (~578:1), 
since a model trained without this would achieve high accuracy by simply 
predicting "not fraud" almost always.

```python
model = LogisticRegression(max_iter=2000,class_weight='balanced')
model.fit(X_train,Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
print(classification_report(Y_test, Y_pred))
```

**Result:** Recall of 0.92 means the model catches 92% of actual fraud cases — 
a strong improvement over an unweighted baseline (which had recall of only 0.58). 
However, precision dropped to 0.06, meaning 94% of transactions flagged as fraud 
are actually legitimate.

## Model 2: Random Forest

Testing whether a tree-based ensemble model handles the class imbalance more 
gracefully than Logistic Regression, while still using `class_weight='balanced'`.

```python
model = RandomForestClassifier(
    n_estimators=211,
    oob_score=True,
    max_depth=4,
    class_weight='balanced',
    random_state=42
)
model.fit(X_train,Y_train)
```

```python
Y_Pred = model.predict(X_test)

print(classification_report(Y_test,Y_Pred))
print(f"OOB Score: {model.oob_score_*100}")
```

**Result:** Random Forest achieves a better balance — recall of 0.89 with 
precision of 0.27 (F1 of 0.41, vs. Logistic Regression's F1 of 0.11). This 
suggests Random Forest is the stronger model for this problem, though precision 
is still low enough to warrant further tuning (see below).

```python
# Threshold tuning: instead of default 0.5 cutoff, test if adjusting the 
# probability threshold improves the precision-recall balance

probs = model.predict_proba(X_test)[:, 1]

for threshold in [0.3, 0.5, 0.7, 0.9]:
    preds = (probs >= threshold).astype(int)
    print(f"--- Threshold: {threshold} ---")
    print(classification_report(Y_test, preds))
```

## Threshold Tuning Conclusion
At threshold 0.9, 
precision improved to 0.84 while recall remained above 0.74, offering a more 
production-realistic balance than the default 0.5 cutoff."]
